# CargoPulse AI — Feature Engineering

## Objective

Transform cleaned historical logistics data into meaningful,
validated features for:

1. Machine Learning
2. Shipment Simulation
3. Business Impact Analysis

## Input Datasets

- DataCo cleaned dataset
- Smart Logistics cleaned dataset

## Important Principles

- Never modify the cleaned source datasets directly.
- Create working copies before transformation.
- Validate every engineered feature.
- Check prediction-time availability.
- Avoid target leakage.
- Do not fabricate missing information.

==========================================Now we will create new columns or features ===============================================

==============================================feature 1(order_hour) ==================================================

```python
dataco_fe["order_hour"] = (
    dataco_fe["order_date_dateorders"].dt.hour
)
```

```python
# print(dataco_fe["order_hour"].value_counts().sort_index())
```

```python
print(
    "DataCo valid hours:",
    dataco_fe["order_hour"].between(0, 23).all()
)
```

```python
print(dataco_fe[
    [
        "order_date_dateorders",
        "shipping_date_dateorders",
        "order_hour"
    ]
].head())
```

```python
print(
    "DataCo valid hours:",
    dataco_fe["order_hour"].between(0, 23).all()
)
```

============================================FEATURE 2(order_day_of_week) ==================================================

```python
dataco_fe["order_day_of_week"] = (
    dataco_fe["order_date_dateorders"].dt.dayofweek
)
```

```python
print("DataCo day_of_week:")
print(
    dataco_fe["order_day_of_week"]
    .value_counts()
    .sort_index()
)
```

```python
print(
    "DataCo valid day_of_week:",
    dataco_fe["order_day_of_week"]
    .between(0, 6)
    .all()
)
```

===============================================feature 3(order_month)=========================================================

```python
dataco_fe["order_month"] =(
    dataco_fe["order_date_dateorders"].dt.month
)
```

```python
print("dataco month")
print(dataco_fe["order_month"].value_counts().sort_index())
```

```python
print(
    "DataCo valid months:",
    dataco_fe["order_month"]
    .between(1, 12)
    .all()
)
```
