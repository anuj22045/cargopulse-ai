==========================================Now we will create new columns or features ===============================================

==============================================feature 1(order_hour) ==================================================

```python
dataco_fe["order_hour"] = (
    dataco_fe["order_date_dateorders"].dt.hour
)
```

```python
# print(dataco_fe["order_hour"].value_counts().sort_index())
```

```python
print(
    "DataCo valid hours:",
    dataco_fe["order_hour"].between(0, 23).all()
)
```

```python
print(dataco_fe[
    [
        "order_date_dateorders",
        "shipping_date_dateorders",
        "order_hour"
    ]
].head())
```

```python
print(
    "DataCo valid hours:",
    dataco_fe["order_hour"].between(0, 23).all()
)
```

============================================FEATURE 2(order_day_of_week) ==================================================

```python
dataco_fe["order_day_of_week"] = (
    dataco_fe["order_date_dateorders"].dt.dayofweek
)
```

```python
print("DataCo day_of_week:")
print(
    dataco_fe["order_day_of_week"]
    .value_counts()
    .sort_index()
)
```

```python
print(
    "DataCo valid day_of_week:",
    dataco_fe["order_day_of_week"]
    .between(0, 6)
    .all()
)
```

===============================================feature 3(order_month)=========================================================

```python
dataco_fe["order_month"] =(
    dataco_fe["order_date_dateorders"].dt.month
)
```

```python
print("dataco month")
print(dataco_fe["order_month"].value_counts().sort_index())
```

```python
print(
    "DataCo valid months:",
    dataco_fe["order_month"]
    .between(1, 12)
    .all()
)
```

=============================================Feature 4(is_weekend)===========================================================

```python
dataco_fe["is_weekend"] = (
    dataco_fe["order_day_of_week"] >= 5
).astype(int)
```

```python
print("DataCo is_weekend:")
print(
    dataco_fe["is_weekend"]
    .value_counts()
    .sort_index()
)
```

```python
print(
    "DataCo valid is_weekend:",
    dataco_fe["is_weekend"]
    .isin([0, 1])
    .all()
)
```

```python
print(
    dataco_fe[
        [
            "order_date_dateorders",
            "order_day_of_week",
            "is_weekend"
        ]
    ].head(20)
)
```

```python
##check if our logic is correct or not  
print(
    pd.crosstab(
        dataco_fe["order_day_of_week"],
        dataco_fe["is_weekend"]
    )
)
```

```python
print("DataCo time features:")
print(
    dataco_fe[
        [
            "order_date_dateorders",
            "order_hour",
            "order_day_of_week",
            "order_month",
            "is_weekend"
        ]
    ].head(10)
)

print("\nDataCo validation:")

print(
    "Hour valid:",
    dataco_fe["order_hour"].between(0, 23).all()
)

print(
    "Day of week valid:",
    dataco_fe["order_day_of_week"].between(0, 6).all()
)

print(
    "Month valid:",
    dataco_fe["order_month"].between(1, 12).all()
)

print(
    "Weekend flag valid:",
    dataco_fe["is_weekend"].isin([0, 1]).all()
)
```

==================================================Shipping Duration features=========================================================

```python
print("Actual Shipping Duration")
print(dataco_fe["days_for_shipping_real"].describe())


print("\nScheduled shipping duration:")
print(dataco_fe["days_for_shipment_scheduled"].describe())

print("Actual shipping duration dtype:")
print(dataco_fe["days_for_shipping_real"].dtype)

print("\nScheduled shipping duration dtype:")
print(dataco_fe["days_for_shipment_scheduled"].dtype)

print("check missing value: ")
print(dataco_fe["days_for_shipping_real"].isna().sum())

print("\n Missing Scheduled shipping duration: ")
print(dataco_fe["days_for_shipment_scheduled"].isna().sum())

print(dataco[["days_for_shipping_real", "days_for_shipment_scheduled"]].head(20))

print(
    "Negative actual durations:",
    (dataco_fe["days_for_shipping_real"] < 0).sum()
)

print(
    "Negative scheduled durations:",
    (dataco_fe["days_for_shipment_scheduled"] < 0).sum()
)
```

```python
#shipping delay features

dataco_fe["shipping_delay_days"] =(
    dataco_fe["days_for_shipping_real"]
    - dataco_fe["days_for_shipment_scheduled"]
)

##negative value means it arrive earlier and positive means it arrive late 
```

```python
print(
    dataco_fe[
        [
            "days_for_shipping_real",
            "days_for_shipment_scheduled",
            "shipping_delay_days"
        ]
    ].head(20)
)
```

```python
#validate the new features

print("Missing shipping_delay_days")
print(dataco_fe["shipping_delay_days"].isna().sum())
```

```python
print(dataco_fe["shipping_delay_days"].describe())
```

```python
validation_difference = (
    dataco_fe["days_for_shipping_real"]
    - dataco_fe["days_for_shipment_scheduled"]
)

print(
    "Calculation correct:",
    dataco_fe["shipping_delay_days"]
    .equals(validation_difference)
)
```

```python
print("Ahead of schedule: ", (dataco_fe["shipping_delay_days"] < 0).sum())

print("Exact Arrival: ", (dataco_fe["shipping_delay_days"] == 0).sum())

print("Late Arrivals: ", (dataco_fe["shipping_delay_days"] > 0).sum())
```

```python
dataco_fe.columns
```

```python
print(dataco_fe.groupby("late_delivery_risk")["shipping_delay_days"].describe())
```
