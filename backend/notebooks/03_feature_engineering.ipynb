```python
# ============================================================
# Step 11 — Historical Performance Feature
# Feature: shipping_mode_delay_rate
# ============================================================

# Create a working copy
dataco_history = dataco_fe.copy()


# ------------------------------------------------------------
# 1. Convert date column to datetime
# ------------------------------------------------------------

dataco_history["order_date_dateorders"] = pd.to_datetime(
    dataco_history["order_date_dateorders"]
)


# ------------------------------------------------------------
# 2. Sort data chronologically
# ------------------------------------------------------------

dataco_history = (
    dataco_history
    .sort_values("order_date_dateorders")
    .reset_index(drop=True)
)


print(
    "Is date sorted:",
    dataco_history["order_date_dateorders"].is_monotonic_increasing
)


# ------------------------------------------------------------
# 3. Calculate statistics for each
#    shipping mode at each timestamp
# ------------------------------------------------------------

mode_time_stats = (
    dataco_history
    .groupby(
        [
            "shipping_mode",
            "order_date_dateorders"
        ]
    )["late_delivery_risk"]
    .agg(
        delay_sum="sum",
        shipment_count="count"
    )
    .reset_index()
)


# ------------------------------------------------------------
# 4. Calculate cumulative historical values
# ------------------------------------------------------------

mode_time_stats["cumulative_delay_sum"] = (
    mode_time_stats
    .groupby("shipping_mode")["delay_sum"]
    .cumsum()
)

mode_time_stats["cumulative_shipment_count"] = (
    mode_time_stats
    .groupby("shipping_mode")["shipment_count"]
    .cumsum()
)


# ------------------------------------------------------------
# 5. Shift by one timestamp
#    This prevents the current timestamp from being used.
# ------------------------------------------------------------

mode_time_stats["previous_delay_sum"] = (
    mode_time_stats
    .groupby("shipping_mode")["cumulative_delay_sum"]
    .shift(1)
)

mode_time_stats["previous_shipment_count"] = (
    mode_time_stats
    .groupby("shipping_mode")["cumulative_shipment_count"]
    .shift(1)
)


# ------------------------------------------------------------
# 6. Create the final feature
# ------------------------------------------------------------

mode_time_stats["shipping_mode_delay_rate"] = (
    mode_time_stats["previous_delay_sum"]
    /
    mode_time_stats["previous_shipment_count"]
)


# ------------------------------------------------------------
# 7. Merge the feature back into the dataset
# ------------------------------------------------------------

dataco_history = dataco_history.merge(
    mode_time_stats[
        [
            "shipping_mode",
            "order_date_dateorders",
            "shipping_mode_delay_rate"
        ]
    ],
    on=[
        "shipping_mode",
        "order_date_dateorders"
    ],
    how="left"
)


# ============================================================
# VALIDATION
# ============================================================

print("\n=== Shipping Mode Delay Rate ===")

print(
    dataco_history[
        [
            "order_date_dateorders",
            "shipping_mode",
            "late_delivery_risk",
            "shipping_mode_delay_rate"
        ]
    ].head(20)
)


# ------------------------------------------------------------
# Check same timestamp consistency
# ------------------------------------------------------------

same_timestamp_variation = (
    dataco_history
    .groupby(
        [
            "shipping_mode",
            "order_date_dateorders"
        ]
    )["shipping_mode_delay_rate"]
    .nunique()
)


print(
    "\nMaximum different historical rates "
    "within the same timestamp:",
    same_timestamp_variation.max()
)


# ------------------------------------------------------------
# Check missing values
# ------------------------------------------------------------

print(
    "\nMissing shipping mode delay rates:",
    dataco_history[
        "shipping_mode_delay_rate"
    ].isna().sum()
)
```
